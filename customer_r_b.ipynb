```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
df=df.rename(columns={'purchase_amount_(usd)':"purchase_amount"})
df.columns
```

```python
df.isnull().sum()
```

```python
df["review_rating"]=df.groupby('category')['review_rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
#create group that in age using "QCUT" funtion in pandas
labels=["young_adult","adult","mid_age","senior"]
df["age_group"]=pd.qcut(df["age"],q=4,labels=labels)
```

```python
df[["age","age_group"]].head(20)
```

```python
# now we need the modified the frequency_of_purchases  using the mapping fun 
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
df["frequency_of_purchases_days"]=df["frequency_of_purchases"].map(frequency_mapping)
```

```python
df[["frequency_of_purchases","frequency_of_purchases_days"]].head(20)
```

```python
(df["discount_applied"]==df["promo_code_used"]).all()
# this mean  both column are in same content
```

```python
df=df.drop("promo_code_used",axis=1)
```
